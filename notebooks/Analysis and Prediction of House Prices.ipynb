<a id="4.-data-preprocessing"></a>

<font size="6.5"><span style="font-family:Altone Trial">4. Data Preprocessing</span></font>

<font size = "5" ><span style= "font-family:Altone Trial">Removing outliers :</span></font>

```python
#Calculate the mean and standard deviation for AveRooms
mean_AveRooms = np.mean(dataset.AveRooms)
sd_AveRooms = np.std(dataset.AveRooms)
mean_AveRooms, sd_AveRooms
```

```python
#Identify points that are more than 3 standard deviations away
three_sd=[rooms for rooms in dataset.AveRooms if (rooms < mean_AveRooms - 3*sd_AveRooms) or (rooms > mean_AveRooms + 3*sd_AveRooms)]
```

```python
#Identify points that are more than 4 standard deviations away
four_sd=[rooms for rooms in dataset.AveRooms if (rooms < mean_AveRooms - 4*sd_AveRooms) or (rooms > mean_AveRooms + 4*sd_AveRooms)]
```

```python
#Identify points that are more than 6 standard deviations away
six_sd=[rooms for rooms in dataset.AveRooms if (rooms < mean_AveRooms - 6*sd_AveRooms) or (rooms > mean_AveRooms + 6*sd_AveRooms)]
```

```python
#Dropping outliers from AveRooms
dataset = dataset[~dataset["AveRooms"].isin(six_sd)]
dataset.shape
```

<font size = "3" ><span style= "font-family:Altone Trial">To better generalize the model that will be built in the later sections- rows where the value of AveRooms that is six standard deviations away from the mean have been dropped.</span></font>

<font size = "3" ><span style= "font-family:Altone Trial">Given the strong relationship between number of rooms and number of bedrooms in an estate (correlation = 0.8), outliers from AveBedrms should have been automatically resolved in the previous step.

<font size = "3" ><span style= "font-family:Altone Trial">We can confirm this by checking the box plot below</span></font>

```python
#Confirming the removal of outliers from AveBedrms 
plt.figure(figsize=(15, 6))
sns.boxplot(x=dataset.AveBedrms, width=0.5, fliersize=3, color='#D7685F');
```

```python
#Calculate the mean and standard deviation for Population
mean_Population = np.mean(dataset.Population)
sd_Population = np.std(dataset.Population)
mean_Population, sd_Population
```

```python
#Identify points that are more than 4 standard deviations away
four_sd_pop=[rooms for rooms in dataset.Population if (rooms < mean_Population - 4*sd_Population) or (rooms > mean_Population + 4*sd_Population)]
```

```python
#Dropping outliers from Population data
dataset = dataset[~dataset["Population"].isin(four_sd_pop)]
dataset.shape
```

<font size = "3" ><span style= "font-family:Altone Trial">To better generalize the model that will be built in the later sections- rows where the value of Population that is four standard deviations away from the mean have been dropped.</span></font>

```python
#Calculate the mean and standard deviation for AveOccup
mean_AveOccup = np.mean(dataset.AveOccup)
sd_AveOccup = np.std(dataset.AveOccup)
mean_AveOccup, sd_AveOccup
```

<font size = "3" ><span style= "font-family:Altone Trial">The values of mean and standard deviation for AveOccup have changed from (3.071, 10.386) to (2.972, 4.323) respectively.


<font size = "3" ><span style= "font-family:Altone Trial">Thus, removal of outliers with resepct to the Population data changed the statistical signature of the AveOccup data.
    
    
<font size = "3" ><span style= "font-family:Altone Trial">This makes sense considering the AvgOccup shows no linear relationship with any other variables except Population (albeit, it's merely 0.1).

```python
#Identify points that are more than 4 standard deviations away
four_sd_occ=[rooms for rooms in dataset.AveOccup if (rooms < mean_AveOccup - 4*sd_AveOccup) or (rooms > mean_AveOccup + 4*sd_AveOccup)]
```

```python
#Identify points that are more than 6 standard deviations away
six_sd_occ=[rooms for rooms in dataset.AveOccup if (rooms < mean_AveOccup - 6*sd_AveOccup) or (rooms > mean_AveOccup + 6*sd_AveOccup)]
```

```python
#Dropping outliers from AveOccup data
dataset = dataset[~dataset["AveOccup"].isin(six_sd_occ)]
dataset.shape
```

<font size = "3" ><span style= "font-family:Altone Trial">Rows where the value of AveOccup that is six standard deviations away from the meaa have been dropped.</span></font>

```python
#Correlation analysis for cleaned dataset
corr = dataset.corr()
plt.figure(figsize=(35,15))
sns.heatmap(corr, cbar=True, square= True, fmt='.1f', annot=True, annot_kws={'size':15}, cmap=sns.light_palette("#D7685F", as_cmap=True));
```

<font size = "5" ><span style= "font-family:Altone Trial">Feature Engineering :

```python
#Bedroom to Room ratio
dataset['BedroomRatio'] = dataset['AveBedrms']/dataset['AveRooms']
dataset.head()
dataset.shape
```

```python
#Rearragning columns
dataset = dataset[['MedInc','HouseAge','AveRooms','AveBedrms','Population','AveOccup','Latitude','Longitude','BedroomRatio','MedHouseVal']]
```

<font size = "5" ><span style= "font-family:Altone Trial">Preparing the data for modelling :

```python
#Creating matrix of independent variables
X = pd.DataFrame(data=dataset.iloc[:, :-1])
print("Matrix of features/independent variables: ", X.shape)
```

```python
#Creating vector of dependent variable
y = pd.DataFrame(data=dataset.iloc[:, -1])
print("Vector of dependent variable: ", y.shape)
```

```python
#Splitting the data into training and test sets
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 1)
```

<font size = "5" ><span style= "font-family:Altone Trial">Feature Scaling :

```python
#Standardisation 
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

<font size = "5" ><span style= "font-family:Altone Trial">Dimensionality Reduction :

<font size = "3" ><span style= "font-family:Altone Trial">We will forego the application of dimensionality reduction since there are a relatively low number of variables / predictors in the dataset.

<a id='5.-model-creation'></a>

<font size = "6.5" ><span style= "font-family:Altone Trial">5. Model Creation</span></font>

<font size = "5" ><span style= "font-family:Altone Trial">Decision Tree :

```python
#Creating object 'dt' of DecisionTreeRegressor
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeRegressor
dt = DecisionTreeRegressor()
```

```python
#Applying k-fold cross validation
R2_dt = cross_val_score(estimator=dt, X=X_train, y=y_train, cv=10, scoring='r2')
print("R^2: {:.2f} %".format(R2_dt.mean()*100))
print("Standard Deviation: {:.2f} %".format(R2_dt.std()*100))
```

<font size = "5" ><span style= "font-family:Altone Trial">Random Forest :

```python
#Creating object 'rf' of RandomForestRegressor
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(n_estimators = 100, random_state = 0)
```

```python
#Applying k-fold cross validation
from sklearn.model_selection import cross_val_score
R2_rf = cross_val_score(estimator=rf, X=X_train, y=y_train.values.ravel(), cv=10, scoring='r2')
print("R^2: {:.2f} %".format(R2_rf.mean()*100))
print("Standard Deviation: {:.2f} %".format(R2_rf.std()*100))
```

<font size = "5" ><span style= "font-family:Altone Trial">XGBoost :

```python
#Creating object 'xgb' of XGBRegressor
from xgboost import XGBRegressor
xgb = XGBRegressor()
```

```python
#Applying k-fold cross validation
R2_xgb = cross_val_score(estimator=xgb, X=X_train, y=y_train, cv=10, scoring='r2')
print("R^2: {:.2f} %".format(R2_xgb.mean()*100))
print("Standard Deviation: {:.2f} %".format(R2_xgb.std()*100))
```

***
<a id='6.-model-evaluation'></a>

<font size = "6.5" ><span style= "font-family:Altone Trial">6. Model Evaluation</span></font>

```python
#Training XGBoost on the Training set
xgb.fit(X_train, y_train)
```

```python
#Predicting the Test set results
y_pred = xgb.predict(X_test)
```

```python
#Evaluating model performance
from sklearn.metrics import r2_score
r2_score(y_test, y_pred)
```

<font size = "3" ><span style= "font-family:Altone Trial">Since there isn't a huge difference in the R2 scores for training and test set, it implies that the model is not overfitting the training data (and is generalizing well to the unseen data in the test set). We will therefore forego the implementation of regularization techniques that are used to address model overfitting.

<font size = "3" ><span style= "font-family:Altone Trial">We can thus conclude that XGBoost performs the best out of all the 3 models and hence will be used for API creation and contanizerising the model using Docker
